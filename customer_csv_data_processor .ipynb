```python
import csv
import json

data = [
    {"name": "John", "age": 30, "city": "New York"},
    {"name": "Jane", "age": 25, "city": "London"},
    {"name": "Bob", "age": 35, "city": "Paris"}
]

with open("people.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "age", "city"])
    writer.writeheader()
    writer.writerows(data)


with open("people.csv", "r", newline="") as f:
    reader = csv.DictReader(f)
    people = list(reader)

    selected_people = []

    for person in people:
        if int(person["age"]) >= 30:
            selected_people.append(person)

    with open("selected_people.json", "w") as f:
        json.dump(selected_people, f, indent=4)

with open("selected_people.json", "r") as f:
    result = json.load(f)
    print("csv -- python -- JSON conversion completed!")
    print("JSON output:")
    print(json.dumps(result, indent=4))
```

```python
import csv

valid = []
invalid = []
other_city = []

with open("customer.csv", "r", newline="") as f:
    reader = csv.DictReader(f)

    for row in reader:
      name = row["name"].strip()
      age_str = row["age"].strip() # Renamed to avoid confusion before conversion
      city = row["city"].strip().lower()

      # Handle empty name, age, or city first
      if name == "" or city == "":
        invalid.append(row)
        continue

      if age_str == "": # Check for empty age string
        invalid.append(row)
        continue

      try:
        age = int(age_str)
      except ValueError:
        invalid.append(row) # If age is not a valid integer, it's invalid
        continue

      # Now apply the filtering logic based on processed data
      if age >= 18 and city == "hyderabad":
        valid.append(row)
      elif age < 18: # If age is less than 18, mark as invalid
        invalid.append(row)
      else: # If age is 18+ but city is not hyderabad
        other_city.append(row)

# Print results after the loop has finished processing all rows
print("valid:", valid)
print("invalid:", invalid)
print("other city:", other_city)
```

```python
def clean_city(city):
  return city.strip().lower()

def process_customer(customer_data):
  age = customer_data["age"]
  city = clean_city(customer_data["city"])

  if age >= 18 and city == "hyderabad":
    return "valid"
  elif age < 18:
    return "invalid"
  else:
    return "other city"

def create_summary(valid_count, invalid_count, other_city_count):
  summary = {
      "valid": valid_count,
      "invalid": invalid_count,
      "other_city": other_city_count
  }
  return summary

def process_customers(customers_list):
  valid_count = 0
  invalid_count = 0
  other_city_count = 0

  for customer_item in customers_list:
    result = process_customer(customer_item)

    if result == "valid": # Corrected to lowercase "valid"
      valid_count += 1
    elif result == "invalid": # Corrected to lowercase "invalid"
      invalid_count += 1
    else:
      other_city_count += 1

  return create_summary(valid_count, invalid_count, other_city_count)

# Redefining 'customer' list for this cell to be self-contained
customer = [
    {"name": "Alice", "age": 32, "city": "hyderabad"},
    {"name": "Bob", "age": 25, "city": "mumbai"},
    {"name": "john", "age": 16, "city": "hyderabad"},
    {"name": "Mike", "age": 35, "city": "hyderabad"}
]

final_report = process_customers(customer)

print(final_report)
```
