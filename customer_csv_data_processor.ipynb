```python
import csv
import json

data = [
    {"name": "John", "age": 30, "city": "New York"},
    {"name": "Jane", "age": 25, "city": "London"},
    {"name": "Bob", "age": 35, "city": "Paris"}
]

with open("people.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "age", "city"])
    writer.writeheader()
    writer.writerows(data)


with open("people.csv", "r", newline="") as f:
    reader = csv.DictReader(f)
    people = list(reader)

    selected_people = []

    for person in people:
        if int(person["age"]) >= 30:
            selected_people.append(person)

    with open("selected_people.json", "w") as f:
        json.dump(selected_people, f, indent=4)

with open("selected_people.json", "r") as f:
    result = json.load(f)
    print("csv -- python -- JSON conversion completed!")
    print("JSON output:")
    print(json.dumps(result, indent=4))
```

```python
import csv
import json

data = [
    {"name": "John", "age": 30, "city": "New York"},
    {"name": "Jane", "age": 25, "city": "London"},
    {"name": "Bob", "age": 35, "city": "Paris"},
    {"name": "Alice", "age": 32, "city": "hyderabad"},
    {"name": "Charlie", "age": 28, "city": "mumbai"}
]

with open("people.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "age", "city"])
    writer.writeheader()
    writer.writerows(data)

with open ("people.csv", "r", newline="") as f:
  reader = csv.DictReader(f)
  customers = list(reader)

  selected_customers = []

  for customer in customers:
    if int(customer["age"]) >= 30 and (customer["city"] == "hyderabad" or \
                                      customer["city"] == "mumbai"):
      selected_customers.append(customer)

  with open("selected_customers.json","w") as f:
    json.dump(selected_customers,f, indent=4)

  with open("selected_customers.json", "r") as f:
    result = json.load(f)
    print("csv -- python -- JSON conversion completed!")
    print("JSON output:")
    print(json.dumps(result, indent=4))

  print("done ! ")
```

```python
import csv
import json

data = [
    {"name": "John", "age": 30, "city": "New York"},
    {"name": "Jane", "age": 25, "city": "London"},
    {"name": "Bob", "age": 35, "city": "Paris"}
]

with open("people.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "age", "city"])
    writer.writeheader()
    writer.writerows(data)

valid_data = []

with open("people.csv", "r", newline="") as f:
    reader = csv.DictReader(f)
    for row in reader:
      name = row["name"].strip()
      age_str = row["age"].strip() # Renamed to avoid confusion before conversion
      city = row["city"].strip().lower()

      if name == "" or age_str == "" or city == "":
        continue

      try:
        age = int(age_str)
      except ValueError:
        continue # Skip if age is not a valid integer

      if age >= 25 and city == "new york":
        customer = {"name": name, "age": age, "city": city}
        valid_data.append(customer)

# Separate the write and read operations
with open("selected_customers.json", "w") as f:
    json.dump(valid_data, f, indent=4)

with open("selected_customers.json", "r") as f:
    result = json.load(f)
    print("csv -- python -- JSON conversion completed!")
    print("JSON output:")
    print(json.dumps(result, indent=4))

print("done ! ")
```

```python
valid = []
invalid = []
other_city = []

age_str = "25"
city = "mumbai"

try:
  age = int(age_str) # Convert age_str to integer here

  if age < 18:
    invalid.append(age)
  elif city.lower() == "hyderabad":
    valid.append(age)
  else:
    other_city.append(age)

except ValueError: # Corrected exception type and indentation
  invalid.append(age_str) # Append the original string if conversion fails, or handle as needed

print("valid:", valid)
print("invalid:", invalid)
print("other city:", other_city)
```

```python
import csv

valid = []
invalid = []
other_city = []

with open("customer.csv", "r", newline="") as f:
    reader = csv.DictReader(f)

    for row in reader:
      name = row["name"].strip()
      age_str = row["age"].strip() # Renamed to avoid confusion before conversion
      city = row["city"].strip().lower()

      # Handle empty name, age, or city first
      if name == "" or city == "":
        invalid.append(row)
        continue

      if age_str == "": # Check for empty age string
        invalid.append(row)
        continue

      try:
        age = int(age_str)
      except ValueError:
        invalid.append(row) # If age is not a valid integer, it's invalid
        continue

      # Now apply the filtering logic based on processed data
      if age >= 18 and city == "hyderabad":
        valid.append(row)
      elif age < 18: # If age is less than 18, mark as invalid
        invalid.append(row)
      else: # If age is 18+ but city is not hyderabad
        other_city.append(row)

# Print results after the loop has finished processing all rows
print("valid:", valid)
print("invalid:", invalid)
print("other city:", other_city)
```
