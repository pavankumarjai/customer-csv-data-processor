# GitHub API Rate Limit Inspector


A Python project to inspect GitHub API rate limits,
convert reset timestamps to IST, and check GitHub usernames.

```python
import requests
from datetime import datetime, timezone, timedelta


def inspect_rate_limit():
    url = "https://api.github.com/users/octocat"

    try:
        response = requests.get(url, timeout=15)

        print("HTTP Status:", response.status_code)
        print("\n--- RATE LIMIT INFORMATION ---")

        headers = response.headers

        total = headers.get("X-RateLimit-Limit")
        remaining = headers.get("X-RateLimit-Remaining")
        reset = headers.get("X-RateLimit-Reset")
        content_type = headers.get("Content-Type")

        print("Total Limit:", total)
        print("Remaining Requests:", remaining)
        print("Content Type:", content_type)

        if reset:
            reset_timestamp = int(reset)

            ist = timezone(timedelta(hours=5, minutes=30))

            reset_time = datetime.fromtimestamp(
                reset_timestamp,
                tz=ist
            )

            print("Reset Time (IST):", reset_time)

        else:
            print("Reset time not available")

    except requests.exceptions.Timeout:
        print("Request timed out")

    except requests.exceptions.RequestException as error:
        print("Network error:", error)

    except (ValueError, TypeError) as error:
        print("Invalid response data:", error)


def check_github_user(username):

    url = f"https://api.github.com/users/{username}"

    try:
        response = requests.get(url, timeout=15)

        print("\n--- USER CHECK ---")
        print("Status:", response.status_code)

        if response.status_code == 200:
            data = response.json()

            print("User Found:", data["login"])
            print("Public Repositories:", data["public_repos"])
            print("Profile URL:", data["html_url"])

        elif response.status_code == 404:
            print("User not found")

        elif response.status_code == 403:
            print("Request forbidden or rate limit exceeded")

        else:
            print("API Error:", response.status_code)

    except requests.exceptions.RequestException as error:
        print("Network error:", error)


# Run the project

inspect_rate_limit()

check_github_user("octocat")

check_github_user("this_user_does_not_exist_123456789")
```

```python
import csv

valid = []
invalid = []
other_city = []

with open("customer.csv", "r", newline="") as f:
    reader = csv.DictReader(f)

    for row in reader:
      name = row["name"].strip()
      age_str = row["age"].strip() # Renamed to avoid confusion before conversion
      city = row["city"].strip().lower()

      # Handle empty name, age, or city first
      if name == "" or city == "":
        invalid.append(row)
        continue

      if age_str == "": # Check for empty age string
        invalid.append(row)
        continue

      try:
        age = int(age_str)
      except ValueError:
        invalid.append(row) # If age is not a valid integer, it's invalid
        continue

      # Now apply the filtering logic based on processed data
      if age >= 18 and city == "hyderabad":
        valid.append(row)
      elif age < 18: # If age is less than 18, mark as invalid
        invalid.append(row)
      else: # If age is 18+ but city is not hyderabad
        other_city.append(row)

# Print results after the loop has finished processing all rows
print("valid:", valid)
print("invalid:", invalid)
print("other city:", other_city)
```

```python
def clean_city(city):
  """Standardizes city names by stripping whitespace and converting to lowercase."""
  return city.strip().lower()

def process_age(age_str):
  """Categorizes an age string into 'missing', 'invalid', 'underage', or 'valid'."""
  age_str = age_str.strip()
  if age_str == "":
    return "missing"
  try:
    age = int(age_str)
  except ValueError:
    return "invalid"
  if age < 18:
    return "underage"
  return "valid"

def clean_customer(customer):
  """Cleans a single customer's data (name, age, city) by stripping and formatting."""
  cleaned = {
      "name": customer["name"].strip().title(),
      "age": str(customer["age"]).strip(), # Ensure age is treated as string for initial processing
      "city": customer["city"].strip().lower()
  }
  return cleaned

def validate_customer(customer_data):
  """Cleans and validates a customer's data, returning age status and cleaned data."""
  cleaned = clean_customer(customer_data)
  age_status = process_age(cleaned["age"])

  cleaned["age_status"] = age_status
  return cleaned

def create_summary(valid_count, invalid_count, other_city_count):
  """Generates a high-level summary of customer counts."""
  summary = {
      "valid": valid_count,
      "invalid": invalid_count,
      "other_city": other_city_count
  }
  return summary

def generate_full_report(customer_data_list, min_age=0, target_city=None):
  """Generates a comprehensive report from a list of raw customer dictionaries.

  The report includes total customers, a high-level summary (using create_summary),
  detailed age status summary, and city-wise customer counts.
  It filters 'valid' customers based on optional min_age and target_city criteria.
  """
  total_customers = len(customer_data_list)

  # These counts will reflect the filtering based on min_age and target_city for 'valid' category
  status_counts = {
      "valid": 0,       # Age >= min_age AND City == target_city
      "underage": 0,    # Age < 18 (from process_age)
      "invalid": 0,     # Invalid age format (from process_age)
      "missing": 0,     # Missing age (from process_age)
      "other_city": 0   # Age >= 18 but doesn't meet report's min_age or target_city
  }
  city_counts = {}

  for customer_item in customer_data_list:
    validated_customer = validate_customer(customer_item)

    customer_age_status = validated_customer["age_status"]
    customer_age_str = validated_customer["age"]
    customer_city = validated_customer["city"]

    # Update status_counts based on report-level min_age and target_city for 'valid' customers
    if customer_age_status == "valid": # This means age >= 18 and is an int
      try:
        actual_age = int(customer_age_str) # Convert to int for comparison

        age_meets_min_age = (actual_age >= min_age)
        city_meets_target_city = (target_city is None or customer_city == target_city.lower())

        if age_meets_min_age and city_meets_target_city:
          status_counts["valid"] += 1
        else:
          # If age is valid (>=18) but does not meet the report's specific min_age or target_city criteria,
          # it falls into 'other_city'.
          status_counts["other_city"] += 1
      except ValueError:
        # This should ideally not happen if process_age returned 'valid'
        status_counts["invalid"] += 1 # Safety net
    elif customer_age_status == "underage":
      status_counts["underage"] += 1
    elif customer_age_status == "invalid":
      status_counts["invalid"] += 1
    elif customer_age_status == "missing":
      status_counts["missing"] += 1

    # Always count city for city_summary, irrespective of filter criteria for age
    city_counts[customer_city] = city_counts.get(customer_city, 0) + 1

  # Aggregate counts for the create_summary function
  high_level_invalid_count = status_counts["underage"] + status_counts["invalid"] + status_counts["missing"]
  high_level_summary = create_summary(status_counts["valid"], high_level_invalid_count, status_counts["other_city"])

  # Construct the full report
  full_report = {
      "total_customers": total_customers,
      "filter_criteria": {
          "min_age": min_age,
          "target_city": target_city
      },
      "high_level_summary": high_level_summary,
      "detailed_age_summary": status_counts, # status_counts already reflects the categories
      "city_summary": city_counts
  }

  return full_report
```

```python
def filter_customers(customers_list, min_age=0, target_city=None):
  result = []
  for customer_item in customers_list:
    try:
      age = int(customer_item["age"])
      if age >= min_age:
        if target_city is None or customer_item["city"].lower() == target_city.lower():
          result.append(customer_item)
    except ValueError:
      pass
  return result

def get_sort_key_ascending(customer_item):
  try:
    age = int(customer_item["age"])
    return age
  except (ValueError, TypeError):
    return float('inf')

def get_sort_key_descending(customer_item):
  try:
    age = int(customer_item["age"])
    return age
  except (ValueError, TypeError):
    return float('-inf')

def process_and_report_customers(customers_list, filter_city=None, minimum_age=0, sort_order=None):
  # Filter customers
  filtered_customers = filter_customers(customers_list, min_age=minimum_age, target_city=filter_city)

  # Sort customers if a sort_order is specified
  if sort_order == "youngest_to_oldest":
    sorted_customers = sorted(filtered_customers, key=get_sort_key_ascending)
  elif sort_order == "oldest_to_youngest":
    sorted_customers = sorted(filtered_customers, key=get_sort_key_descending, reverse=True)
  else:
    sorted_customers = filtered_customers # No specific sort order

  total_selected = len(sorted_customers)

  report = {
      "filter_city": filter_city,
      "minimum_age": minimum_age,
      "sort_order": sort_order,
      "total_selected": total_selected,
      "customers": sorted_customers
  }
  return report

# Define the customer list (from previous cell context)
customer = [
    {"name": "Alice", "age": "32", "city": "Hyderabad"},
    {"name": "Bob", "age": "25", "city": "Mumbai"},
    {"name": "john", "age": "abc", "city": "Hyderabad"},
    {"name": "Sara", "age": " ", "city": "Delhi"},
    {"name": "mike", "age": "35", "city": "Hyderabad"},
    {"name": "ravi", "age": "xyz", "city": "Hyderabad"},
    {"name": "Sita", "age": " ", "city": "mumbai"}
]

# Example Usage 1: Filter Hyderabad customers, min age 30, sorted youngest to oldest
report_1 = process_and_report_customers(
    customer, filter_city="Hyderabad", minimum_age=30, sort_order="youngest_to_oldest"
)
print("\nReport for Hyderabad (min age 30, youngest to oldest):\n", report_1)

# Example Usage 2: Filter Mumbai customers, min age 20, sorted oldest to youngest
report_2 = process_and_report_customers(
    customer, filter_city="Mumbai", minimum_age=20, sort_order="oldest_to_youngest"
)
print("\nReport for Mumbai (min age 20, oldest to youngest):\n", report_2)

# Example Usage 3: Filter all customers, min age 18, no specific sort
report_3 = process_and_report_customers(
    customer, minimum_age=18, sort_order=None
)
print("\nReport for all customers (min age 18, no sort):\n", report_3)
```

```python
import requests

url = "https://api.github.com/users/octocat"
response = requests.get(url)

print("Status:", response.status_code)
print("Total limit:", response.headers.get("X-RateLimit-Limit"))
print("Remaining:", response.headers.get("X-RateLimit-Remaining"))
print("Reset:", response.headers.get("X-RateLimit-Reset"))
print("Content-Type:", response.headers.get("Content-Type"))
```

```python
from datetime import datetime, timezone

reset_timestamp = int(response.headers.get("X-RateLimit-Reset"))

reset_time = datetime.fromtimestamp(
    reset_timestamp,
    tz=timezone.utc
)

print("Rate limit resets at:", reset_time)
```

```python
import requests

def check_github_user(username):
    url = f"https://api.github.com/users/{username}"

    response = requests.get(url)

    print("Status:", response.status_code)

    if response.status_code == 200:
        data = response.json()
        print("User found:", data["login"])
    elif response.status_code == 404:
        print("User not found!")
    else:
        print("API error:", response.status_code)

check_github_user("octocat")
check_github_user("this_user_does_not_exist_123456789")
```
