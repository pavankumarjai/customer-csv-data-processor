```python

import requests

username = input("Enter GitHub username: ")

url = f"https://api.github.com/users/{username}"

try:
    response = requests.get(url, timeout=10)
    response.raise_for_status()

    data = response.json()

    print("\n--- GitHub User Details ---")
    print("Username:", data.get("login"))
    print("Followers:", data.get("followers"))
    print("Public Repositories:", data.get("public_repos"))

except requests.exceptions.Timeout:
    print("Request timed out. Try again.")

except requests.exceptions.HTTPError:
    print("HTTP Error:", response.status_code)

except requests.exceptions.RequestException as e:
    print("Request failed:", e)
```

```python
import requests
import time

def get_github_user(username):
    max_attempts = 3

    for attempt in range(1, max_attempts + 1):
        try:
            print(f"Attempt {attempt}")

            response = requests.get(
                f"https://api.github.com/users/{username}",
                timeout=10
            )

            response.raise_for_status()

            data = response.json()

            print("Username:", data["login"])
            print("Followers:", data["followers"])

            return data

        except requests.exceptions.Timeout:
            print("Request timed out")

        except requests.exceptions.ConnectionError:
            print("Connection error")

        except requests.exceptions.HTTPError as error:
            print("HTTP Error:", error)
            return None

        if attempt < max_attempts:
            wait_time = 2 ** attempt
            print(f"Waiting {wait_time} seconds...")
            time.sleep(wait_time)

    print("Maximum attempts reached.")
    return None

get_github_user("octocat")
```

```python
import logging
import requests
import time

# Defining the function directly here to ensure it is defined and self-contained
def get_github_user(username):
    max_attempts = 3
    for attempt in range(1, max_attempts + 1):
        try:
            response = requests.get(f"https://api.github.com/users/{username}", timeout=10)
            response.raise_for_status()
            return response.json()
        except requests.exceptions.Timeout:
            logging.warning("Request timed out. Retrying...")
        except requests.exceptions.ConnectionError:
            logging.warning("Connection error. Retrying...")
        except requests.exceptions.HTTPError as error:
            logging.error(f"HTTP Error: {error}")
            return None
        if attempt < max_attempts:
            wait_time = 2 ** attempt
            time.sleep(wait_time)
    return None

logging.info("GitHub User Checker started")
username = "octocat"
logging.info(f"Checking GitHub user: {username}")

result = get_github_user(username)

if result:
    logging.info("GitHub user fetched successfully")
else:
    logging.error("Failed to fetch GitHub user")
```
